```python
import pandas as pd
import numpy as np

%matplotlib inline
```

```python
equipos = ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C', 'D', 'D', 'D']
goles = [4, 2, 0, 3, 3, 2, 1, 1, 0, 2, 4, 0]

datos = pd.DataFrame({'Equipos': equipos, 'Goles': goles})

datos
```

```python
datos.plot(kind= 'scatter',
           y= 'Goles',
           x= 'Equipos')
```

```python
datos.plot(kind= 'scatter',
           y= 'Goles',
           x= 'Equipos',
           title= 'Goles por equipo',
           xlabel= 'Equipos participantes',
           ylabel= 'Goles realizados')
```

```python
datos.plot(kind= 'scatter',
           y= 'Goles',
           x= 'Equipos',
           title= 'Goles por equipo',
           xlabel= 'Equipos participantes',
           ylabel= 'Goles realizados',
           s= 100,
           c= 'red')
```
