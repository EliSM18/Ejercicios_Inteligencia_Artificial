```python
import pandas as pd
import numpy as np
```

```python
# Promedio de mujeres
mu = 169
# Desviación estándar
sigma = 5
# Cantidad
cantidad = 250
# Iniciando el generador
np.random.seed(0)
# Generación de datos
altura_mujer  = np.random.normal(mu, sigma, cantidad).astype(int)

# Promedio de hombres
mu = 176
# Desviación estándar
sigma = 6
# Cantidad
muestra = 250
# Iniciando el generador
np.random.seed(1)
# Generación de datos
altura_hombre  = np.random.normal(mu, sigma, muestra).astype(int)
```

```python
altura_mujer
```

```python
altura_hombre
```

```python
# Uniendo valores
estatura = pd.DataFrame({'Mujeres': altura_mujer, 'Hombres': altura_hombre})
estatura
```

```python
estatura.plot(kind= 'hist')
```

```python
estatura.plot(kind= 'hist',
              grid= True,
              bins= 20,
              column= 'Hombres',
              ylabel= 'Frecuencia',
              xlabel= 'Centímetros',
              title= 'Histograma de la estatura de los hombres')
```

```python
# Asignando el resultado a un objeto
grafico = estatura.plot(kind= 'hist',
                        grid= True,
                        bins= 20,
                        column= 'Hombres',
                        title= 'Histograma de la estatura de los hombres')

grafico.set_ylabel('Frecuencia')
grafico.set_xlabel('Centímetros')
```

```python
# Asignando el resultado a un objeto
grafico = estatura.plot(kind= 'hist',
                        grid= True,
                        bins= 20,
                        title= 'Histograma de la estatura de los hombres',
                        alpha= 0.6)

grafico.set_ylabel('Frecuencia')
grafico.set_xlabel('Centímetros')
```

```python
estatura.groupby('Mujeres').count().plot(kind= 'bar')                                     
```
