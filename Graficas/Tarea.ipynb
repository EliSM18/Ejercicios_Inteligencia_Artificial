```python
import pandas as pd
%matplotlib inline

df = pd.read_csv("../CSVs/preciosCasas.csv")

print(df)
```

```python
# Porcentaje del número de dormitorios de las casas
df = df[df['bedrooms'] > 0]
conteo = df['bedrooms'].value_counts()

conteo.plot(
    kind='pie',  # Tipo de gráfico: barras
    title='Distribución de Dormitorios',  # Título del gráfico
    autopct= '%1.0f%%'
)
```

```python
# Los 10 precios más altos de las casas
precios_altos = df['price'].nlargest(10)
precios_altos
```

```python
precios_altos.plot(
    kind= 'bar',
    grid= True,
    title= 'Los 10 precios más altos de las casas',
    ylabel= 'Precio',
    xlabel= 'Casas'
);
```

```python
# Asegurarse que no se incluyan 0s en el análisis de precios
df = df[df['price'] > 0]

#  Los 10 precios más bajos de las casas
precios_bajos = df['price'].nsmallest(10)
precios_bajos
```

```python
precios_bajos.plot(
    kind= 'bar',
    grid= True,
    title= 'Los 10 precios más bajos de las casas',
    ylabel= 'Precio',
    xlabel= 'Casas'
);
```
