```python
from scipy import sparse

# Crear matriz
eye = np.eye(4)
print("NumPy array:\n{}".format(eye))
```

```python
sparse_matrix = sparse.csr_matrix(eye)
print("\nSiPy sparse CRS matrix:\n{}".format(sparse_matrix))
```

```python
import pandas as pd

data = {'Nombre':["Juan", "Ana", "Pedro", "Rosa"],
        'Localidad':["Veracruz", "Colima", "Yucatán", "Sonora"],
        'Edad':[24, 13, 53, 33]
        }
data_pandas = pd.DataFrame(data)

display(data_pandas)
```

```python
# Seleccionar mayor a 30
display(data_pandas[data_pandas.Edad > 30])
```
