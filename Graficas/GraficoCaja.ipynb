```python
import pandas as pd
import numpy as np
```

```python
# Promedio de mujeres
mu = 169
# Desviación estándar
sigma = 5
# Cantidad
cantidad = 250
# Iniciando el generador
np.random.seed(0)
# Generación de datos
altura_mujer  = np.random.normal(mu, sigma, cantidad).astype(int)

# Promedio de hombres
mu = 176
# Desviación estándar
sigma = 6
# Cantidad
muestra = 250
# Iniciando el generador
np.random.seed(1)
# Generación de datos
altura_hombre  = np.random.normal(mu, sigma, muestra).astype(int)
```

```python
altura_mujer
```

```python
altura_hombre
```

```python
estatura = pd.DataFrame({'Mujeres': altura_mujer, 'Hombres': altura_hombre})
estatura
```

```python
estatura.plot(kind= 'box',
              ylabel= 'Centímetros',
              title= 'Distribución de las estaturas de mujeres y hombres')
```

```python
grafico = estatura.plot(kind= 'box',
                        grid= True,
                        ylabel= 'Centímetros',
                        title= 'Distribución de las estaturas de mujeres y hombres',
                        column= ['Hombres', 'Mujeres'])
grafico.set_xlabel('Sexo')
```
