```python
import pandas as pd
import numpy as np

%matplotlib inline
```

```python
# Equipos
equipos = ['A', 'A', 'A', 'B', 'B', 'B', 'C', 'C', 'C', 'D', 'D', 'D']
goles = [4, 2, 0, 3, 3, 2, 1, 1, 0, 2, 4, 0]

datos = pd.DataFrame({'Equipos': equipos, 'Goles': goles})

datos
```

```python
agrupaDatos = datos.groupby(by=[equipos]).sum()
agrupaDatos
```

```python
agrupaDatos.plot(kind= 'pie', y= 'Goles')
```

```python
datos.groupby(by=[equipos]).sum().plot(kind= 'pie',
                                       y= 'Goles',
                                       autopct= '%1.0f%%',
                                       title= 'Porcentaje de goles por cada equipo durante el torneo',
                                       ylabel= 'Goles totales')
```

```python
colores = ['lightblue', 'lightpink', 'lightyellow', 'lightgreen']

datos.groupby(by=[equipos]).sum().plot(kind= 'pie', 
                                       y= 'Goles',
                                       autopct= '%1.0f%%',
                                       title= 'Porcentaje de goles por cada equipo durante el torneo',
                                       ylabel= 'Goles totales',
                                       colors= colores)
```
