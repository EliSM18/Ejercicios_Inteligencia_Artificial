```python
import pandas as pd
import numpy as np

%matplotlib inline
```

```python
x = [2001, 2002, 2003, 2004, 2005, 2006]

# y con distintos ingresos
trabajo = [4, 11, 12, 9, 10, 8]
inversiones = [1, 1.5, 2.5, 2.5, 3.5, 3]
freelance = [2, 4, 10, 6, 10, 11]
youtube = [1, 1.1, 1.2, 0.9, 1.4, 1.8]
```

```python
datos = pd.DataFrame({'Sueldo': trabajo, 'Inversiones': inversiones, 'Freelancing': freelance, 'Redes': youtube}, x)

datos
```

```python
# Gráfica de área
datos.plot(kind= 'area',
           grid= True,
           title= 'Distribución del ingreso en los diferentes años',
           ylabel= 'Miles Dlls',
           xlabel= 'Años',
)
```

```python
datos.plot(kind= 'area',
           grid= True,
           title= 'Distribución del ingreso en los diferentes años',
           ylabel= 'Miles Dlls',
           xlabel= 'Años',
           stacked= False, # evitar que las gráficas se apilen
           alpha= 0.7      # modificar la transparenia
)
```
