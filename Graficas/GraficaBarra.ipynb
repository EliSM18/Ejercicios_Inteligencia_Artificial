```python
import pandas as pd
import numpy as np

%matplotlib inline
```

```python
# Eje x años
x = [2001, 2002, 2003, 2004, 2005, 2006]

# Ganancias en diccionario
y = {'Ganancia': [4, 11, 12, 9, 10, 8], 'Inverión': [6, 8, 8, 6, 7, 5]}

# Lista a dataframe
datos = pd.DataFrame(y, x)

# Visualizar datos
datos
```

```python
# Gráfica de Líneas
datos.plot(kind= 'line',
           grid= True,
           title= 'Variación de las ganancias',
           ylabel= 'miles Dlls',
           xlabel= 'Años',
           xlim= (2001, 2006),
           ylim= (0, 12.5));

```

```python
#Gráfica de barras
datos.plot(kind= 'bar',
           grid= True,
           title= 'Variación de las ganancias',
           ylabel= 'miles Dlls',
           xlabel= 'Años',
           xlim= (2001, 2006),
           ylim= (0, 12.5));
```

```python
#Gráfica de barras encimadas
datos.plot(kind= 'bar',
           grid= True,
           stacked= True,
           title= 'Variación de las ganancias y las inversiones',
           ylabel= 'miles Dlls',
           xlabel= 'Años',
           xlim= (2001, 2006));
```

```python
#Gráfica de barras horizontales
datos.plot(kind= 'barh',
           grid= True,
           stacked= True,
           title= 'Variación de las ganancias y las inversiones',
           ylabel= 'miles Dlls',
           xlabel= 'Años');
```
