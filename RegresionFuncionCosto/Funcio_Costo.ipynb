```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('../CSVs/preciosCasas.csv')

x = df.iloc[0:10,4].values
y = df.iloc[0:10,1].values

print(x)
print(y)
```

```python

# Inicializar los escaladores
mms_x = MinMaxScaler()
mms_y = MinMaxScaler()

# Ajustar y transformar los datos
X_std = mms_x.fit_transform(x.reshape(-1,1))
Y_std = mms_y.fit_transform(y.reshape(-1,1))

print(X_std)
print(Y_std)
```

```python
# Graficar ecuaciones
y_p1 = X_std - 2
y_p2 = 2 * X_std + 5
y_p3 = 0.1 * X_std + 0.05
y_p4 = X_std + 2

fig, axs = plt.subplots(1,4, figsize=(15,4))

axs[0].scatter(X_std, Y_std, color='orange')
axs[0].plot(X_std, y_p1, color='black')
axs[0].set_title('y = x - 2')


axs[1].scatter(X_std, Y_std, color='gray')
axs[1].plot(X_std, y_p2, color='black')
axs[1].set_title('y = 2x + 5')

axs[2].scatter(X_std, Y_std, color='green')
axs[2].plot(X_std, y_p3, color='black')
axs[2].set_title('y = 0.1x + 0.05')

axs[3].scatter(X_std, Y_std, color='red')
axs[3].plot(X_std, y_p4, color='black')
axs[3].set_title('y = x + 2')

plt.show()
```
