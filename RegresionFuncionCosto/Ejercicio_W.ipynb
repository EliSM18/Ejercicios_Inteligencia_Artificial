```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
W = [-0.5, 0.0, 0.5, 1.0, 1.5, 2.0, 2.5]
X = [1.0, 2.0, 3.0]
Y = [1.0, 2.0, 3.0]
M = 3

resultados = []

for w in W:
    sumatoria = 0
    for x in X:
        res = np.power((w*x)-x, 2)
        sumatoria = sumatoria + res

    resultado = sumatoria / (2 *  M)
    resultados.append(resultado)

print(resultados)
```

```python
plt.scatter(W, resultados, color='blue', label='Datos')
plt.plot(W, resultados, color='red', label='Curva')
plt.title('Función de Costo')
plt.xlabel('Valores w')
plt.ylabel('Resultados obtenidos')
plt.xticks(W)
plt.legend()
plt.show()
```
