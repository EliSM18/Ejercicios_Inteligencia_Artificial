```python
import numpy as np

xi = [1.75,1.84,1.95,1.62,1.79,1.82,1.64,1.72,1.89]
yi = [11,3,0,2,5,9,16,8,9]

media_xi = np.mean(xi)
media_yi = np.mean(yi)

xi_menos_media = xi - media_xi

yi_menos_media = yi - media_yi

multiplicacion = xi_menos_media * yi_menos_media

suma_multi = np.sum(multiplicacion)

cuadrado = np.power(xi_menos_media, 2)

suma_cuadrado = np.sum(cuadrado)

m = suma_multi / suma_cuadrado

c = media_yi - (m * media_xi)

print("m =",m, "\nc =",c)
```

```python
print("==== CANTIDAD DE GOLES ===")
print("Predice la cantidad de goles de acuerdo a la estatura\n")

x_entrada = input("Introduce la cantidad de faltas ")

x_entrada = float(x_entrada)

y = (m * x_entrada) + c

print("Goles=", y)
```
