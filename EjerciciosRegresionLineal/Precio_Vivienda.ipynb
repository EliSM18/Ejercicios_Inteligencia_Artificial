```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

print(f"x_train: {x_train}")
print(f"y_train: {y_train}")
```

```python
print(f"x_train.shape: {x_train.shape}")

m = x_train.shape[0]

print(f"El número de muestras de entrenamiento es: {m}")
```

```python
m = len(x_train)
print(f"El número de muestras de entrenamiento es: {m}")
```

```python
# Entrenamiento
i = 0

x_i = x_train[i]
y_i = y_train[i]

print(f"(x^({i}), y^({i})) = ({x_i}, {y_i})")
```

```python
#  Trazando los datos

plt.scatter(x_train, y_train, marker='+', c='r')
plt.title("Precios de la vivienda")
plt.ylabel("Precios (en 1000s de Dlls)")
plt.xlabel("Tamaño(en 1000s pies cuadrados)")
plt.show()
```

```python
w = 100
b = 100

print(f"w: {w}")
print(f"b: {b}")
```

```python
def compute_model_output(x, w, b):
    """
    Calcula la predicción de un modelo lineal
    Args:
      x (ndarray (m,)): Datos de entrada
      w, b (scalar): Parámetros del modelo
    Returns:
        f_wb (ndarray (m,)): Predicciones del modelo
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
tmp_f_wb = compute_model_output(x_train, w, b)

plt.plot(x_train, tmp_f_wb, c='b', label='Nuestra predicción')

plt.scatter(x_train, y_train, marker='x', c='r', label='Valores reales')
plt.title("Predicciones de la vivienda")
plt.ylabel("Precio (en 1000s de Dlls)")
plt.xlabel("Tamaño (en 1000s de pies cuadrados)")
plt.legend()
plt.show()
```
