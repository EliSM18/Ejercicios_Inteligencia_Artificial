```python
import numpy as np

xi = [12,3,0,2,9,4,7,20,5,8]
yi = [2,9,14,12,11,5,5,1,4,6]

media_xi = np.mean(xi)
media_yi = np.mean(yi)

xi_menos_media = xi - media_xi

yi_menos_media = yi - media_yi

multiplicacion = xi_menos_media * yi_menos_media

suma_multi = np.sum(multiplicacion)

cuadrado = np.power(xi_menos_media, 2)

suma_cuadrado = np.sum(cuadrado)

m = suma_multi / suma_cuadrado

c = media_yi - (m * media_xi)

print("m =",m, "\nc =",c)
```

```python
print("==== CANTIDAD DE ERRORES ===")
print("Predice la cantidad de meses en los que puede suceder ese número de errores\n")

x_entrada = input("Introduce la cantidad de equivocaciones ")

x_entrada = float(x_entrada)

y = (m * x_entrada) + c

print("Meses=", y)
```
